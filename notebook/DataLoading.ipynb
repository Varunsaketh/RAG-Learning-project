```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../Data/textfiles/pythonintro.txt",encoding = "utf-8")
textdoc1 = loader.load()
print(textdoc1)
```

```python
from langchain_community.document_loaders import DirectoryLoader
drloader = DirectoryLoader(
    "../Data/textfiles", # need to provide the path of the directory where our text files are stored
    glob = "**/*.txt", # glob pattern is used to specify the types of files we want to load in this case we are loading all text files in the directory and its subdirectories
    loader_cls = TextLoader, # we are specifying the loader class to use for loading the documents in this case we are using TextLoader which is a built in loader for loading text files
    loader_kwargs = {"encoding": "utf-8"}
)
docs = drloader.load()
print(len(docs))
print(docs)
print(docs[0])
print(docs[1])
```

```python
from langchain_community.document_loaders import PyPDFLoader , PyMuPDFLoader
drloader = DirectoryLoader(
    "../Data/pdf_files",
    glob ="**/*.pdf",  # we use ** before the file name so it will extract pdfs also from the hidden subfolders not only fro the folder we assigned
    loader_cls = PyMuPDFLoader
)
pdfdoc1 = drloader.load()
print(pdfdoc1)
```
